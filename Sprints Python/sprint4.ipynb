## Ejercicio 7.9 · Traza de ejecución de un agente

```python
traza = {
    "paso": "agente_principal", "tipo": "llm", "tokens": 1200,
    "hijos": [
        {"paso": "buscar_clientes", "tipo": "tool", "herramienta": "sql", "tokens": 0, "hijos": []},
        {"paso": "subagente_analista", "tipo": "llm", "tokens": 2300, "hijos": [
            {"paso": "calcular_churn", "tipo": "tool", "herramienta": "python", "tokens": 0, "hijos": []},
            {"paso": "buscar_contexto", "tipo": "tool", "herramienta": "rag", "tokens": 450, "hijos": [
                {"paso": "reescribir_consulta", "tipo": "llm", "tokens": 300, "hijos": []},
            ]},
        ]},
        {"paso": "redactar_respuesta", "tipo": "llm", "tokens": 1800, "hijos": []},
    ],
}

def tokens_totales(nodo):
    return nodo.get('tokens', 0) + sum(tokens_totales(h) for h in nodo.get('hijos', []))

def profundidad(nodo):
    if not nodo.get('hijos'):
        return 1
    return 1 + max(profundidad(h) for h in nodo.get('hijos'))

def herramientas(nodo):
    h_set = {nodo['herramienta']} if nodo.get('tipo') == 'tool' and 'herramienta' in nodo else set()
    for h in nodo.get('hijos', []):
        h_set.update(herramientas(h))
    return h_set

def llamadas_llm(nodo):
    count = 1 if nodo.get('tipo') == 'llm' else 0
    return count + sum(llamadas_llm(h) for h in nodo.get('hijos', []))

print(f"Tokens totales: {tokens_totales(traza)}")
print(f"Profundidad máxima: {profundidad(traza)}")
print(f"Herramientas usadas: {sorted(list(herramientas(traza)))}")
print(f"Llamadas al LLM: {llamadas_llm(traza)}")

# El subagente es el segundo hijo (índice 1)
subagente = traza["hijos"][1]
print(f"Tokens del subagente: {tokens_totales(subagente)}")
```

## Ejercicio 7.10 · Resumen jerárquico de documentos largos

```python
def resumir(textos):
    return " ".join([" ".join(t.split()[:3]) for t in textos])

capitulos = [
    "Ventas crecen un 12% en el canal online",
    "Tiendas físicas estables respecto al trimestre anterior",
    "Devoluciones bajan tras el cambio de política",
    "Coste logístico sube por el precio del combustible",
    "Nuevo almacén en Valencia reduce plazos de entrega",
]

def resumen_jerarquico(textos, grupo=2, nivel=0):
    print(f"Nivel {nivel}: {len(textos)} textos")
    
    # Condición base: si solo queda un texto, se devuelve
    if len(textos) == 1:
        return textos[0]
        
    # Paso recursivo: agrupar y resumir
    grupos = [textos[i:i+grupo] for i in range(0, len(textos), grupo)]
    resumenes = [resumir(g) for g in grupos]
    
    return resumen_jerarquico(resumenes, grupo=grupo, nivel=nivel+1)

resultado_g2 = resumen_jerarquico(capitulos, grupo=2)
print(f"Resumen final: {resultado_g2}\n")

resultado_g3 = resumen_jerarquico(capitulos, grupo=3)
print(f"Resumen final: {resultado_g3}")
```
