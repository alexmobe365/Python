## Ejercicio 5.1 · Proyección y filtro de registros

A partir de una lista de diccionarios:

1. Crear una lista que conserve únicamente `id` e `importe` de los registros cuyo importe sea superior a 100.
2. Calcular la suma de esos importes.
3. Obtener los canales distintos y mostrarlos ordenados.

```python
registros = [
    {"id": 1, "cliente": "Acme", "importe": 250.0, "canal": "web"},
    {"id": 2, "cliente": "Globex", "importe": 80.0, "canal": "tienda"},
    {"id": 3, "cliente": "Initech", "importe": 1200.0, "canal": "web"},
    {"id": 4, "cliente": "Umbrella", "importe": 99.99, "canal": "app"},
    {"id": 5, "cliente": "Acme", "importe": 430.0, "canal": "app"},
]

# 1. Filtrar los registros y proyectar únicamente las claves necesarias.
seleccionados = [
    {"id": registro["id"], "importe": registro["importe"]}
    for registro in registros
    if registro["importe"] > 100
]

# 2. Sumar los importes de los registros seleccionados.
total = sum(registro["importe"] for registro in seleccionados)

# 3. Obtener los canales sin duplicados y ordenarlos.
canales = sorted({registro["canal"] for registro in registros})

print(seleccionados)
print(f"Total: {total}")
print(f"Canales: {canales}")
```

### Explicación

- La comprehension de `seleccionados` recorre cada diccionario de `registros`.
- La condición `if registro["importe"] > 100` descarta los importes menores o iguales a 100.
- El diccionario creado dentro de la comprehension incluye solo las claves `id` e `importe`: esto es la **proyección**.
- `sum()` suma los importes de la lista resultante.
- `{registro["canal"] for registro in registros}` crea un conjunto (`set`), por lo que elimina duplicados. `sorted()` devuelve una lista ordenada alfabéticamente.

## Ejercicio 5.2 · Transponer filas a columnas

Tenemos una matriz de 3 filas y 4 columnas. Cada fila representa un registro con los campos `id`, `unidades`, `precio` y `descuento`.

Construir una lista por cada columna mediante una comprehension anidada, sin utilizar `zip()`. Después, sumar las unidades, que corresponden a la columna de índice 1.

```python
# 3 filas x 4 columnas: id, unidades, precio y descuento.
filas = [
    [1, 10, 2.5, 0.0],
    [2, 4, 10.0, 0.1],
    [3, 7, 3.2, 0.0],
]

# La comprehension exterior recorre los índices de las columnas.
# La interior recoge, para cada columna, el valor de todas las filas.
columnas = [
    [fila[indice_columna] for fila in filas]
    for indice_columna in range(len(filas[0]))
]

# La columna 1 contiene las unidades.
unidades_totales = sum(columnas[1])

print(columnas)
print(f"Unidades totales: {unidades_totales}")
```

### Explicación

La comprehension tiene dos niveles:

1. `for indice_columna in range(len(filas[0]))` recorre los índices de las cuatro columnas: `0`, `1`, `2` y `3`.
2. `[fila[indice_columna] for fila in filas]` construye una lista con los valores de esa columna, recorriendo todas las filas.

Así se transforma la matriz de filas a columnas sin utilizar `zip()`. Como las unidades están en la columna de índice `1`, se suman con `sum(columnas[1])`.

## Ejercicio 5.3 · Inferir el esquema a partir de un registro de muestra

A partir de un diccionario de ejemplo:

1. Construir un diccionario que relacione cada columna con el nombre del tipo de su valor.
2. Crear una lista con las columnas cuyo valor sea `None`, ya que su tipo real no puede inferirse a partir de ese valor.

```python
muestra = {
    "id": 1043,
    "cliente": "Acme",
    "importe": 99.9,
    "activo": True,
    "fecha": "2026-10-01",
    "descuento": None
}

# Diccionario: nombre de columna -> nombre del tipo de dato.
esquema = {
    columna: type(valor).__name__
    for columna, valor in muestra.items()
}

# Columnas cuyo valor es None y, por tanto, no permiten inferir el tipo real.
columnas_sin_tipo = [
    columna
    for columna, valor in muestra.items()
    if valor is None
]

print(esquema)
print(f"No se puede inferir: {columnas_sin_tipo}")
```

### Explicación

- `muestra.items()` permite recorrer simultáneamente cada clave (`columna`) y su valor (`valor`).
- En la comprehension de diccionario, `type(valor).__name__` obtiene el nombre del tipo de Python: por ejemplo, `int`, `str`, `float` o `bool`.
- `None` tiene el tipo `NoneType`, pero ese valor no revela cuál sería el tipo de dato esperado cuando la columna tenga información. Por eso, la segunda comprehension selecciona las columnas cuyo valor es `None`.

## Ejercicio 5.4 · Generar las rutas de particiones

Generar todas las rutas posibles combinando el año, los meses y los días indicados. El formato debe ser:

`ventas/year=AAAA/month=MM/day=DD`

El mes y el día deben aparecer siempre con dos dígitos.

```python
anio = 2026
meses = [9, 10]
dias = [1, 2, 3]

# Una comprehension anidada genera todas las combinaciones de mes y día.
rutas = [
    f"ventas/year={anio}/month={mes:02d}/day={dia:02d}"
    for mes in meses
    for dia in dias
]

# Mostrar cada ruta y el número total de particiones.
for ruta in rutas:
    print(ruta)

print(f"Particiones: {len(rutas)}")
```

### Explicación

- La comprehension recorre primero cada mes de `meses` y, para cada mes, recorre todos los días de `dias`.
- El f-string inserta el año, el mes y el día en la ruta.
- El formato `:02d` indica que el número debe mostrarse como entero (`d`) con un mínimo de dos posiciones, rellenando con un cero a la izquierda cuando sea necesario. Por ejemplo, `9` se convierte en `09`.
- Se generan `2 × 3 = 6` rutas, una por cada combinación de mes y día.
- `len(rutas)` devuelve el número total de particiones generadas.
