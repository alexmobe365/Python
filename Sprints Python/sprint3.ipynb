## Ejercicio 5.1 · Proyección y filtro de registros

```python
registros = [
    {"id": 1, "cliente": "Acme", "importe": 250.0, "canal": "web"},
    {"id": 2, "cliente": "Globex", "importe": 80.0, "canal": "tienda"},
    {"id": 3, "cliente": "Initech", "importe": 1200.0, "canal": "web"},
    {"id": 4, "cliente": "Umbrella", "importe": 99.99, "canal": "app"},
    {"id": 5, "cliente": "Acme", "importe": 430.0, "canal": "app"},
]

# 1. Lista filtrada
filtrados = [{"id": r["id"], "importe": r["importe"]} for r in registros if r["importe"] > 100]
print(filtrados)

# 2. Total de los importes
total = sum(r["importe"] for r in filtrados)
print(f"Total: {total}")

# 3. Canales distintos ordenados
canales = sorted({r["canal"] for r in registros})
print(f"Canales: {canales}")
```

## Ejercicio 5.2 · Transponer filas a columnas

```python
# 3 filas x 4 columnas (id, unidades, precio, descuento)
filas = [
    [1, 10, 2.5, 0.0],
    [2, 4, 10.0, 0.1],
    [3, 7, 3.2, 0.0],
]

# Transponer con comprehension anidada
columnas = [[fila[i] for fila in filas] for i in range(len(filas[0]))]
print(columnas)

# Sumar unidades totales (columna 1)
unidades_totales = sum(columnas[1])
print(f"Unidades totales: {unidades_totales}")
```

## Ejercicio 5.3 · Inferir el esquema a partir de un registro de muestra

```python
muestra = {"id": 1043, "cliente": "Acme", "importe": 99.9, "activo": True,
           "fecha": "2026-10-01", "descuento": None}

# Comprehension de diccionario
esquema = {k: type(v).__name__ for k, v in muestra.items()}
print(esquema)

# Obtener campos no inferibles
no_inferibles = [k for k, v in esquema.items() if v == 'NoneType']
print(f"No se puede inferir: {no_inferibles}")
```

## Ejercicio 5.4 · Generar las rutas de particiones

```python
anio = 2026
meses = [9, 10]
dias = [1, 2, 3]

# Generar rutas
rutas = [f"ventas/year={anio}/month={m:02d}/day={d:02d}" for m in meses for d in dias]

for ruta in rutas:
    print(ruta)

print(f"Particiones: {len(rutas)}")
```
